# SQL Important Cheat Sheet (Page 1/15)

## 1. SQL Fundamentals

### Database
- Organized collection of related data.

### DBMS
- Software used to store, manage, and retrieve data.

### RDBMS
- Data stored in related tables.

### SQL
- Structured Query Language, used to communicate with relational databases.

### Core Terms
- **Table:** Collection of related data in rows and columns.
- **Row:** A single record in a table.
- **Column:** A single attribute or field in a table.
- **Schema:** Database structure (tables, columns, relationships, etc.).

### SQL Command Categories
- **DDL (Data Definition Language):** Defines and modifies database structure. Examples: `CREATE`, `ALTER`, `DROP`, `TRUNCATE`
- **DML (Data Manipulation Language):** Manipulates data in tables. Examples: `INSERT`, `UPDATE`, `DELETE`
- **DQL (Data Query Language):** Retrieves data from tables. Examples: `SELECT`
- **DCL (Data Control Language):** Controls access to data. Examples: `GRANT`, `REVOKE`
- **TCL (Transaction Control Language):** Manages transactions. Examples: `COMMIT`, `ROLLBACK`, `SAVEPOINT`

### Working of SQL Query
SQL Query → RDBMS (Process Query) → Result (Data / Output)

### Common Databases
- MySQL
- PostgreSQL
- SQL Server
- Oracle
- SQLite

### Key Takeaway
- **SQL is declarative:** It describes what data you need, rather than telling the database step-by-step how to do it.
